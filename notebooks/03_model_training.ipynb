# Transfer-learning training
The shared training entry point supports ResNet50, EfficientNet-B0 and B3. Both use GAP in torchvision and a 3-logit head; softmax is used for inference confidence.

```python
# From the project root:
# !python train.py --model resnet50 --imbalance-strategy sampler
# !python train.py --model efficientnet_b0 --imbalance-strategy weighted_loss
# !python train.py --model resnet50 --use-roi-annotations --roi-mode mask
```

Training saves the best validation-loss checkpoint, JSON history, and loss/accuracy curves in `models/<model>/`.
